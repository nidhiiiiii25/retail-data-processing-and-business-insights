```python
def profile_dataset(df, name):

    print(f"\n{'='*50}")
    print(f"DATASET: {name}")
    print(f"{'='*50}")

    print(f"Shape: {df.shape[0]} rows × {df.shape[1]} columns")

    print("\n--- Missing Values ---")
    nulls = df.isnull().sum()
    null_pct = (nulls / len(df) * 100).round(2)

    print(
        pd.DataFrame({
            'missing_count': nulls,
            'missing_%': null_pct
        })[nulls > 0]
    )

    print("\n--- Duplicates ---")
    print(f"Fully duplicate rows: {df.duplicated().sum()}")

    if 'transaction_id' in df.columns:
        print(
            f"Duplicate transaction_ids: "
            f"{df['transaction_id'].duplicated().sum()}"
        )

    print("\n--- Data Types ---")
    print(df.dtypes)

    print("\n--- Categorical Columns ---")

    categorical_cols = [
        'category',
        'product_name',
        'purchase_location',
        'payment_method',
        'payment_status'
    ]

    for col in categorical_cols:
        if col in df.columns:
            print(f"\n{col}")
            print(f"Unique Values: {df[col].nunique()}")
            print(df[col].unique()[:10])

    print("\n--- Numeric Summary ---")

    numeric_cols = [
        col for col in
        ['price', 'quantity', 'discount']
        if col in df.columns
    ]

    print(df[numeric_cols].describe())
    
profile_dataset(retail_data1, "Retail Data 1")
```
